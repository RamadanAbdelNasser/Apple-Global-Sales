```python
Avg_Customer_Ratings = df[df['has_rating']==True]['customer_rating'].mean()
Avg_Customer_Ratings 
```

```python
Total_Revenue = df['revenue_usd'].sum()
Total_Revenue
```

```python
Total_Revenue_across_Years_df = df.groupby('year')['revenue_usd'].sum().to_frame().reset_index()
```

```python
apple_minimal_colors = [
    "#000000",  # Black
    "#1C1C1E",  # Dark Gray / almost black
    "#2C2C2E",  # Dark Gray
    "#3A3A3C",  # Medium Dark Gray
    "#48484A",  # Medium Gray
    "#636366",  # Gray
    "#8E8E93",  # Light Gray
    "#AEAEB2",  # Lighter Gray
    "#C7C7CC",  # Light Gray
    "#D1D1D6",  # Very Light Gray
    "#0071E3",  # Apple Blue Highlight
    "#34C759",  # Apple Green Highlight
]
```

```python
sales_distribution_cross_years_fig = px.line(Total_Revenue_across_Years_df,'year','revenue_usd')
sales_distribution_cross_years_fig.update_xaxes(dtick=1)
sales_distribution_cross_years_fig.update_layout(
    paper_bgcolor="#FFFFFF",  
    plot_bgcolor="#FFFFFF",
    margin=dict(l=20, r=20, t=20, b=20),  # حواف داخلية
    shapes=[
        dict(
            type="rect",
            xref="paper",
            yref="paper",
            x0=0,
            y0=0,
            x1=1,
            y1=1,
            line=dict(color="black", width=2),
            fillcolor="rgba(0,0,0,0)"  # شفاف
        )
    ]
    # bordercolor="Black",
)

sales_distribution_cross_years_fig.show()
```
